```python
import tiktoken

tokenizer = tiktoken.get_encoding("gpt2")
```

```python
text = ("""Well!--even through the prism of Hermia's tears I felt able to

face the fact with equanimity. Poor Jack Gisburn! The women had

made him--it was fitting that they should mourn him.""")

integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
integers
```

```python
output = tokenizer.decode(integers)
output
```
